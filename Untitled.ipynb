```python
import torch
import torchaudio
from torchaudio.pipelines import HDEMUCS_HIGH_MUSDB_PLUS
from torchaudio.transforms import Fade
import os
from basic_pitch.inference import predict_and_save, Model
from basic_pitch import ICASSP_2022_MODEL_PATH

# ---------------- Demucs Separation ----------------
def separate_sources(model, mix, bundle, segment=10., overlap=0.1, device=None):
    if device is None:
        device = mix.device
    else:
        device = torch.device(device)

    batch, channels, length = mix.shape
    chunk_len = int(bundle.sample_rate * segment * (1 + overlap))
    start = 0
    end = chunk_len
    overlap_frames = int(overlap * bundle.sample_rate)

    fade = Fade(fade_in_len=0, fade_out_len=overlap_frames, fade_shape='linear')
    final = torch.zeros(batch, len(model.sources), channels, length, device=device)

    while start < length - overlap_frames:
        chunk = mix[:, :, start:end]
        with torch.no_grad():
            out = model.forward(chunk)
        out = fade(out)
        final[:, :, :, start:end] += out
        if start == 0:
            fade.fade_in_len = overlap_frames
        start += chunk_len - overlap_frames
        end += chunk_len - overlap_frames
        if end >= length:
            fade.fade_out_len = 0

    return final

# ---------------- Paths ----------------
input_file = "037_Ed_Sheeran_-_Shape_Of_You_Midi_Culture_Remix_(mp3.pm).wav"
demucs_output_dir = "separated_audio"
midi_output_dir = "midi_output"
os.makedirs(demucs_output_dir, exist_ok=True)
os.makedirs(midi_output_dir, exist_ok=True)

# ---------------- Load Audio ----------------
input_wav, sample_rate = torchaudio.load(input_file)

# ---------------- Load Demucs Model ----------------
bundle = HDEMUCS_HIGH_MUSDB_PLUS
demucs_model = bundle.get_model()
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
demucs_model.to(device)

# ---------------- Normalize Audio ----------------
input_wav = input_wav.to(device)
input_wav = torchaudio.functional.resample(input_wav, sample_rate, bundle.sample_rate)
ref = input_wav.mean(0)
input_wav = (input_wav - ref.mean()) / ref.std()

# ---------------- Separate Sources ----------------
sources = separate_sources(demucs_model, input_wav[None], bundle, device=device)[0]
sources = sources * ref.std() + ref.mean()
source_names = demucs_model.sources
separated_sources = {name: source for name, source in zip(source_names, sources)}

# ---------------- Save Demucs Stems ----------------
for name, source_wav in separated_sources.items():
    wav_path = os.path.join(demucs_output_dir, f"{name}.wav")
    torchaudio.save(wav_path, source_wav.cpu(), bundle.sample_rate)
    print(f"Saved {wav_path}")

# ---------------- Load Basic Pitch Model ----------------
bp_model = Model(ICASSP_2022_MODEL_PATH)

# ---------------- Convert Stems to MIDI ----------------
for name in source_names:
    wav_path = os.path.join(demucs_output_dir, f"{name}.wav")
    midi_path = os.path.join(midi_output_dir, f"{name}.mid")

    # Basic Pitch will save MIDI automatically
    predict_and_save(
        [wav_path],
        midi_output_dir,
        save_midi=True,
        sonify_midi=False,       # change to True if you want MIDI audio as WAV
        save_model_outputs=False,
        save_notes=False,
        model_path=ICASSP_2022_MODEL_PATH
    )
    print(f"Converted {wav_path} to MIDI at {midi_path}")
```
